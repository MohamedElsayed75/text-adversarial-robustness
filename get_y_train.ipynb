```python
import pandas as pd

# 1. Load the data
df_full = pd.read_csv('IMDB Dataset.csv')
y_test = pd.read_csv('y_test.csv')

# 2. Define the mapping
# positive -> 1, negative -> 0
sentiment_map = {'positive': 1, 'negative': 0}

# 3. Extract y_train by dropping test indices
# This assumes y_test.csv shares the same index as df_full
train_df = df_full.drop(y_test.index)

# 4. Convert and Extract y_train
y_train = train_df['sentiment'].map(sentiment_map)

# 5. Save the result
y_train.to_csv('y_train.csv', index=False)

print("Conversion successful!")
print(y_train.value_counts()) # Check the 0s and 1s distribution
```
